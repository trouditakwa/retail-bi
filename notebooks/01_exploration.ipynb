```python
import pandas as pd
from pathlib import Path
```

```python
file_path = Path("../data/raw/Online Retail.xlsx")

print("fichier trouvé:", file_path.exists())
print("chemin utilisé:", file_path.resolve())
```

```python
df=pd.read_excel(file_path)
print("donnees chargées avec succés !")
```

```python
print("Nombre de lignes:", df.shape[0])
print("Nombre de colonnes:", df.shape[1])
```

```python
print("Nom des colonnes:", df.columns.tolist())
```

```python
df.info()
```

```python
df.describe(include='all').T  #afficher les statistiques descriptives pour toutes les colonnes, y compris les colonnes non numériques
```

```python
df.isna().sum().sort_values(ascending=False)  #compter le nombre de valeurs manquantes pour chaque colonne et trier les résultats par ordre décroissant
```

```python
print("Quantités négatives:", (df['Quantity'] < 0).sum())
print("Prix négatifs:", (df['UnitPrice'] < 0).sum())
```

```python
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

cancelled = df[df["InvoiceNo"].str.startswith("C")]
print("Nombre de factures annulées:", len(cancelled))
cancelled.head()    #identifier les colonnes annulées en utilisant la méthode str.startswith() pour vérifier si le numéro de facture commence par "C".
```
